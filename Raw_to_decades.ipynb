```python
import os
import shutil
import re

raw_books_path = "../Datasets/fr_books_raw"
organized_folders = "../Datasets/decade_grouped_corpus2"

os.makedirs(organized_folders, exist_ok=True)

# Just a counter to see if we're actually doing something
processed_count = 0

for book_name in os.listdir(raw_books_path):
    if not book_name.endswith(".txt"):
        continue
    
    # Find the year
    found_year = re.search(r"\d{4}", book_name)
    
    if found_year:
        year_val = int(found_year.group())
        decade_label = f"{(year_val // 10) * 10}s"
        
        target_dir = os.path.join(organized_folders, decade_label)
        os.makedirs(target_dir, exist_ok=True)
        
        shutil.copy(
            os.path.join(raw_books_path, book_name),
            os.path.join(target_dir, book_name)
        )
        processed_count += 1

print(f"Done! We dorted {processed_count} books into their decades.")
```

```python
from pathlib import Path

root_dir = Path("../Datasets/decade_grouped_corpus2")
merged_dir = Path("../Datasets/decade_merged2")
merged_dir.mkdir(parents=True, exist_ok=True)

# Sort the decades (so we can process 1810 before 1820 for instance)
for folder in sorted([d for d in root_dir.iterdir() if d.is_dir()]):
    output_filename = merged_dir / f"{folder.name}.txt"
    
    print(f"Merging {folder.name}...")
    
    with output_filename.open("w", encoding="utf-8") as big_file:
        # Sort files inside the decade too
        for book_file in sorted(folder.glob("*.txt")):
            # Added some space so words from book A don't accidentally merge with words from book B
            big_file.write("\n\n") 
            
            # Using 'errors=ignore' because some books may have weird symbols that break the utf-8 reader
            with book_file.open("r", encoding="utf-8", errors="ignore") as f:
                for line in f:
                    big_file.write(line)

print("Merging finished. Now we have one big file per decade!")
```
