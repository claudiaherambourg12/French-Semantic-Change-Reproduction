```python
import pandas as pd

# Load the cleaned data
df = pd.read_csv('../Datasets/final_results_CLEANED.csv')

top_shifters = df.groupby('word').agg({
    'semantic_change': 'mean',
    'frequency': 'mean',
    'polysemy': 'mean'
}).reset_index()

top_shifters = top_shifters.sort_values(by='semantic_change', ascending=False)

print("--- TOP 10 MOST CHANGED WORDS (FRENCH CORPUS) ---")
print(top_shifters.head(10))

top_shifters.to_csv('../Datasets/top_semantic_shifters.csv', index=False)
```
