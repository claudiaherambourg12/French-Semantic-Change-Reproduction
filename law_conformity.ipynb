```python
import numpy as np
import os
import json
import pandas as pd
from scipy.spatial.distance import cosine

aligned_dir = '../Datasets/decade_aligned'
vocab_dir = '../Datasets/decade_stats'
output_file = '../Datasets/semantic_change_results.csv'

files = sorted([f for f in os.listdir(aligned_dir) if f.endswith("-aligned.npy")])
results = []

for i in range(len(files) - 1):
    # Decade T
    dec1_name = files[i].replace("-aligned.npy", "")
    vecs1 = np.load(os.path.join(aligned_dir, files[i]))
    with open(os.path.join(vocab_dir, f"{dec1_name}_vocab.txt"), "r", encoding='utf-8') as f:
        vocab1 = [line.strip() for line in f]
    with open(os.path.join(vocab_dir, f"{dec1_name}_counts.json"), "r", encoding='utf-8') as f:
        counts1 = json.load(f)
        
    # Decade t+1
    dec2_name = files[i+1].replace("-aligned.npy", "")
    vecs2 = np.load(os.path.join(aligned_dir, files[i+1]))
    with open(os.path.join(vocab_dir, f"{dec2_name}_vocab.txt"), "r", encoding='utf-8') as f:
        vocab2 = [line.strip() for line in f]

    # Map vocab to indices
    idx1 = {w: i for i, w in enumerate(vocab1)}
    idx2 = {w: i for i, w in enumerate(vocab2)}

    # Compare every word found in both decades
    common_vocab = set(vocab1).intersection(set(vocab2))
    print(f"Comparing {dec1_name} and {dec2_name}: {len(common_vocab)} shared words.")

    for word in common_vocab:
        v1 = vecs1[idx1[word]]
        v2 = vecs2[idx2[word]]
        
        # Calculate volatility (cosine distance)
        change = cosine(v1, v2)
        
        results.append({
            'word': word,
            'decade_start': dec1_name,
            'frequency': counts1[word],
            'semantic_change': change
        })
df = pd.DataFrame(results)
df.to_csv(output_file, index=False)
print(f"Analysis complete! Results saved to {output_file}")
```
