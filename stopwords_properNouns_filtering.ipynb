```python
import pandas as pd
import spacy
import numpy as np

nlp = spacy.load("fr_core_news_sm")
df = pd.read_csv('../Datasets/final_laws_results.csv')
valid_pos = {'NOUN', 'VERB', 'ADJ', 'ADV'}

def is_content_word(word):
    doc = nlp(word)
    if len(doc) == 0:
        return False
    
    token = doc[0]
    
    # remove stop words
    if token.is_stop:
        return False
    
    # remove proper nouns
    if token.pos_ == 'PROPN':
        return False
    
    if token.pos_ in valid_pos:
        return True
    
    return False

print("Applying final Hamilton et al. filters...")
unique_words = df['word'].unique()
valid_map = {word: is_content_word(word) for word in unique_words}

df_to_a_t = df[df['word'].map(valid_map)].copy()

df_to_a_t.to_csv('../Datasets/final_results_CLEANED.csv', index=False)

print(f"Original entries: {len(df)}")
print(f"Entries after filtering: {len(df_to_a_t)}")
```
