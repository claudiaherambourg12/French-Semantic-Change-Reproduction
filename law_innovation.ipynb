```python
import os
import pandas as pd
import numpy as np
from sklearn.neighbors import NearestNeighbors

aligned_dir = '../Datasets/decade_aligned'
vocab_dir = '../Datasets/decade_stats'
output_file = '../Datasets/final_laws_results.csv'
K = 20


results_df = pd.read_csv('../Datasets/semantic_change_results.csv')
polysemy_data = []

files = sorted([f for f in os.listdir(aligned_dir) if f.endswith("-aligned.npy")])

for filename in files:
    decade = filename.replace("-aligned.npy", "")
    print(f"Calculating Polysemy for: {decade}...")
    
    vecs = np.load(os.path.join(aligned_dir, filename))
    with open(os.path.join(vocab_dir, f"{decade}_vocab.txt"), "r", encoding='utf-8') as f:
        vocab = [line.strip() for line in f]
    
    # Use nearestneighbors and metric='cosine' finds the neighbors with the smallest cosine distance
    nn = NearestNeighbors(n_neighbors=K+1, metric='cosine', algorithm='brute')
    nn.fit(vecs)
    
    # Find distances to the nearest k+1 neighbors (including self)
    distances, indices = nn.kneighbors(vecs)
    
    for i, word in enumerate(vocab):
        mean_cosine_dist = np.mean(distances[i][1:]) 
        
        polysemy_data.append({
            'word': word,
            'decade_start': decade,
            'polysemy': mean_cosine_dist})

#Merge
poly_df = pd.DataFrame(polysemy_data)
final_df = pd.merge(results_df, poly_df, on=['word', 'decade_start'])


final_df.to_csv(output_file, index=False)
print(f"Success! Polysemy calculated and merged.")
```
