```python
import numpy as np
import os
from scipy.sparse import load_npz
from sklearn.utils.extmath import randomized_svd
import pickle


input_dir = '../Datasets/decade_matrices'
output_dir = '../Datasets/decade_svd_300'
os.makedirs(output_dir, exist_ok=True)
DIM = 300
N_ITER = 5 # As seen in their code: n_iter=5

for filename in sorted(os.listdir(input_dir)):
    if filename.endswith("_ppmi.npz"):
        decade = filename.replace("_ppmi.npz", "")
        print(f"Running Randomized SVD for {decade}...")

        # PPMI matrix
        matrix = load_npz(os.path.join(input_dir, filename))

        # Replicate randomized_svd from their script
        u, s, v = randomized_svd(matrix, n_components=DIM, n_iter=N_ITER, random_state=42)

        # Save as individual components exactly like their code
        np.save(os.path.join(output_dir, f"{decade}-u.npy"), u)
        np.save(os.path.join(output_dir, f"{decade}-s.npy"), s)
        np.save(os.path.join(output_dir, f"{decade}-v.npy"), v)
        
        print(f"Successfully saved SVD components for {decade}")

print("SVD complete for all decades.")
```
