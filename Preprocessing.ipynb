```python
import os
import re

source_dir = "../Datasets/decade_merged"
output_dir = "../Datasets/decade_cleaned"

if not os.path.exists(output_dir):
    os.makedirs(output_dir)

for filename in os.listdir(source_dir):
    if filename.endswith(".txt"):
        print(f"Cleaning {filename}...")
        
        with open(os.path.join(source_dir, filename), 'r', encoding='utf-8') as f:
            # Lowercase (Paper Appendix A) 
            text = f.read().lower()
            
            # Strip Punctuation (Paper Appendix A) 
            # We replace punctuation with a space so words don't accidentally glue together and this handles en-dash (–), quotes (« »), and apostrophes
            text = re.sub(r'[^\w\s]', ' ', text)
            
            # Remove digits
            text = re.sub(r'\d+', ' ', text)
            
            # 4. Whitespace cleanup (this turns all tabs, newlines, and multiple spaces into a single space)
            clean_text = " ".join(text.split())

            with open(os.path.join(output_dir, filename), 'w', encoding='utf-8') as out_f:
                out_f.write(clean_text)

print("Pre-processing complete!")
```
