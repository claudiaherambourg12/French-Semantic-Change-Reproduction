```python
import collections
import json
from pathlib import Path


input_folder = Path("../Datasets/decade_cleaned")

stats_output = Path("../Datasets/decade_stats")
stats_output.mkdir(parents=True, exist_ok=True)

# We follow the study with a threshold of 100 
freq_threshold = 100 


for txt_file in sorted(input_folder.glob("*.txt")):
    decade_name = txt_file.stem 
    
    with txt_file.open("r", encoding="utf-8") as f:
        all_tokens = f.read().split()
        
    token_counts = collections.Counter(all_tokens)
    
    # This is the direct input for the Law of Conformity (f)
    reliable_counts = {word: count for word, count in token_counts.items() if count >= freq_threshold}
    
    # Save the counts as JSON (need these later for thefFrequency (f) variable in the regression)
    json_path = stats_output / f"{decade_name}_counts.json"
    with open(json_path, "w", encoding="utf-8") as j_file:
        json.dump(reliable_counts, j_file)
        
    # Vocab list (use this list to define the dimensions of our PPMI matrix)
    sorted_vocab = sorted(reliable_counts.keys(), key=lambda x: reliable_counts[x], reverse=True)
    
    vocab_path = stats_output / f"{decade_name}_vocab.txt"
    with open(vocab_path, "w", encoding="utf-8") as v_file:
        for word in sorted_vocab:
            v_file.write(f"{word}\n")

    print(f"Done! {decade_name} has {len(sorted_vocab)} in the final vocabulary.")
```
